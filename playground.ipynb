# "execute" Playground

> Test the context of the Minecraft "execute" command chain.

- Designed to have the same precision error as in-game.
- Context forks, such as "as" are not implemented.
- The Execute class itself acts as an entity.
- Even if you input x/z coordinates as integers, they are assigned as .0, not 0.5.

```python
import numpy as np
from execute import Execute
```

```python
p1 = Execute().positioned("0 0 0").rotated("0 0")
p2 = Execute().positioned("-1 0 1").rotated("0 0")
p1.facing(p2)

print(f"pos: {p1.position}, rot: {p1.rotation}")
```

```python
A = Execute().facing("~1 ~ ~1")
print(f"[A] pos: {A.position}, rot: {A.rotation}")

marker = Execute().positioned("~1 ~ ~1")
B = Execute().facing(marker).facing("^ ^ ^1").facing("^ ^ ^1")
print(f"[B] pos: {B.position}, rot: {B.rotation}")
```
